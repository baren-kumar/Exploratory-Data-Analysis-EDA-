```python
# 1. Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 3. Load Dataset

df = pd.read_csv("https://raw.githubusercontent.com/baren-kumar/Exploratory-Data-Analysis-EDA-/refs/heads/main/advanced_business_management_dataset.csv")
```

```python
# 4. Display Dataset
# ============================================================

print("Dataset Shape:", df.shape)
display(df.head())

print("\nColumn Names:")
print(df.columns)
```

```python
# 5. Basic Dataset Information
# ============================================================

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
# 6. Convert Date Column
# ============================================================

df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.to_period("M").astype(str)
```

```python
# 7. KPI Summary
# ============================================================

total_units_sold = df["Units_Sold"].sum()
total_gross_sales = df["Gross_Sales"].sum()
total_discount = df["Discount_Amount"].sum()
total_net_sales = df["Net_Sales"].sum()
total_cost = df["Total_Cost"].sum()
total_profit = df["Profit"].sum()
average_profit_margin = df["Profit_Margin"].mean() * 100
average_customer_rating = df["Customer_Rating"].mean()
average_delivery_time = df["Delivery_Time_Days"].mean()

print("========== KPI SUMMARY ==========")
print("Total Units Sold:", total_units_sold)
print("Total Gross Sales:", round(total_gross_sales, 2))
print("Total Discount Amount:", round(total_discount, 2))
print("Total Net Sales:", round(total_net_sales, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))
print("Average Profit Margin: {:.2f}%".format(average_profit_margin))
print("Average Customer Rating: {:.2f}".format(average_customer_rating))
print("Average Delivery Time: {:.2f} days".format(average_delivery_time))
print("=================================")
```

```python
# 8. Monthly Sales and Profit Analysis
# ============================================================

monthly_summary = df.groupby("Month").agg({
    "Units_Sold": "sum",
    "Gross_Sales": "sum",
    "Net_Sales": "sum",
    "Total_Cost": "sum",
    "Profit": "sum",
    "Order_ID": "count"
}).reset_index()

monthly_summary.rename(columns={"Order_ID": "Total_Orders"}, inplace=True)

monthly_summary["Net_Sales_Growth_%"] = monthly_summary["Net_Sales"].pct_change() * 100

display(monthly_summary)
plt.figure(figsize=(12,5))
plt.plot(monthly_summary["Month"], monthly_summary["Net_Sales"], marker="o")
plt.title("Monthly Net Sales Trend")
plt.xlabel("Month")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

plt.figure(figsize=(12,5))
plt.plot(monthly_summary["Month"], monthly_summary["Profit"], marker="o")
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

plt.figure(figsize=(12,5))
plt.bar(monthly_summary["Month"], monthly_summary["Net_Sales_Growth_%"])
plt.title("Monthly Net Sales Growth %")
plt.xlabel("Month")
plt.ylabel("Growth %")
plt.xticks(rotation=45)
plt.show()
```

```python
# 9. Region-wise Analysis
# ============================================================

region_summary = df.groupby("Region").agg({
    "Net_Sales": "sum",
    "Profit": "sum",
    "Units_Sold": "sum",
    "Order_ID": "count",
    "Customer_Rating": "mean"
}).reset_index()

region_summary.rename(columns={"Order_ID": "Total_Orders"}, inplace=True)

display(region_summary)

plt.figure(figsize=(10,5))
plt.bar(region_summary["Region"], region_summary["Net_Sales"])
plt.title("Region-wise Net Sales")
plt.xlabel("Region")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10,5))
plt.bar(region_summary["Region"], region_summary["Profit"])
plt.title("Region-wise Profit")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
# 10. Product-wise Analysis
# ============================================================

product_summary = df.groupby("Product").agg({
    "Units_Sold": "sum",
    "Net_Sales": "sum",
    "Profit": "sum"
}).reset_index()

product_summary = product_summary.sort_values(by="Net_Sales", ascending=False)

display(product_summary)

plt.figure(figsize=(10,5))
plt.bar(product_summary["Product"], product_summary["Net_Sales"])
plt.title("Product-wise Net Sales")
plt.xlabel("Product")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10,5))
plt.bar(product_summary["Product"], product_summary["Profit"])
plt.title("Product-wise Profit")
plt.xlabel("Product")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
# 11. Category-wise Analysis
# ============================================================

category_summary = df.groupby("Category").agg({
    "Net_Sales": "sum",
    "Profit": "sum",
    "Units_Sold": "sum"
}).reset_index()

display(category_summary)

plt.figure(figsize=(7,7))
plt.pie(
    category_summary["Net_Sales"],
    labels=category_summary["Category"],
    autopct="%1.1f%%"
)
plt.title("Category-wise Net Sales Contribution")
plt.show()
```

```python
# 12. Profit and Loss Analysis
# ============================================================

plt.figure(figsize=(12,5))
plt.plot(monthly_summary["Month"], monthly_summary["Net_Sales"], marker="o", label="Net Sales")
plt.plot(monthly_summary["Month"], monthly_summary["Total_Cost"], marker="o", label="Total Cost")
plt.plot(monthly_summary["Month"], monthly_summary["Profit"], marker="o", label="Profit")
plt.title("Monthly Net Sales, Cost, and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 13. Profit Margin Analysis
# ============================================================

profit_margin_product = df.groupby("Product")["Profit_Margin"].mean().reset_index()
profit_margin_product["Profit_Margin_%"] = profit_margin_product["Profit_Margin"] * 100

display(profit_margin_product)

plt.figure(figsize=(10,5))
plt.bar(profit_margin_product["Product"], profit_margin_product["Profit_Margin_%"])
plt.title("Average Profit Margin by Product")
plt.xlabel("Product")
plt.ylabel("Profit Margin %")
plt.xticks(rotation=45)
plt.show()
```

```python
# 14. Customer Type Analysis
# ============================================================

customer_type_summary = df.groupby("Customer_Type").agg({
    "Net_Sales": "sum",
    "Profit": "sum",
    "Customer_ID": "nunique",
    "Customer_Rating": "mean"
}).reset_index()

customer_type_summary.rename(columns={"Customer_ID": "Unique_Customers"}, inplace=True)

display(customer_type_summary)

plt.figure(figsize=(7,7))
plt.pie(
    customer_type_summary["Net_Sales"],
    labels=customer_type_summary["Customer_Type"],
    autopct="%1.1f%%"
    )
plt.title("Net Sales by Customer Type")
plt.show()

plt.figure(figsize=(8,5))
plt.bar(customer_type_summary["Customer_Type"], customer_type_summary["Customer_Rating"])
plt.title("Average Customer Rating by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Average Rating")
plt.show()
```

```python
# 15. Marketing Channel Analysis
# ============================================================

marketing_summary = df.groupby("Marketing_Channel").agg({
    "Net_Sales": "sum",
    "Profit": "sum",
    "Order_ID": "count"
}).reset_index()

marketing_summary.rename(columns={"Order_ID": "Total_Orders"}, inplace=True)

display(marketing_summary)

plt.figure(figsize=(10,5))
plt.bar(marketing_summary["Marketing_Channel"], marketing_summary["Net_Sales"])
plt.title("Marketing Channel-wise Net Sales")
plt.xlabel("Marketing Channel")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# 16. Inventory Status Analysis
# ============================================================

inventory_summary = df.groupby("Product").agg({
    "Inventory_Level": "mean",
    "Units_Sold": "sum"
}).reset_index()

display(inventory_summary)

plt.figure(figsize=(10,5))
plt.bar(inventory_summary["Product"], inventory_summary["Inventory_Level"])
plt.title("Average Inventory Level by Product")
plt.xlabel("Product")
plt.ylabel("Average Inventory Level")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(8,5))
plt.scatter(df["Inventory_Level"], df["Units_Sold"])
plt.title("Inventory Level vs Units Sold")
plt.xlabel("Inventory Level")
plt.ylabel("Units Sold")
plt.grid(True)
plt.show()
```

```python
# 17. Delivery Performance Analysis
# ============================================================

supplier_delivery = df.groupby("Supplier").agg({
    "Delivery_Time_Days": "mean",
    "Profit": "sum",
    "Net_Sales": "sum"
}).reset_index()

display(supplier_delivery)
plt.figure(figsize=(10,5))
plt.bar(supplier_delivery["Supplier"], supplier_delivery["Delivery_Time_Days"])
plt.title("Average Delivery Time by Supplier")
plt.xlabel("Supplier")
plt.ylabel("Delivery Time Days")
plt.xticks(rotation=45)
plt.show()
```

```python
# 18. Salesperson / Employee Performance
# ============================================================

salesperson_summary = df.groupby("Salesperson").agg({
    "Net_Sales": "sum",
    "Profit": "sum",
    "Units_Sold": "sum",
    "Order_ID": "count",
    "Customer_Rating": "mean"
}).reset_index()

salesperson_summary.rename(columns={"Order_ID": "Total_Orders"}, inplace=True)
salesperson_summary = salesperson_summary.sort_values(by="Net_Sales", ascending=False)

display(salesperson_summary)
plt.figure(figsize=(10,5))
plt.bar(salesperson_summary["Salesperson"], salesperson_summary["Net_Sales"])
plt.title("Salesperson-wise Net Sales")
plt.xlabel("Salesperson")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
# 19. Discount Analysis
# ============================================================

plt.figure(figsize=(8,5))
plt.scatter(df["Discount_Percent"], df["Net_Sales"])
plt.title("Discount Percent vs Net Sales")
plt.xlabel("Discount Percent")
plt.ylabel("Net Sales")
plt.grid(True)
plt.show()

plt.figure(figsize=(8,5))
plt.scatter(df["Discount_Percent"], df["Profit"])
plt.title("Discount Percent vs Profit")
plt.xlabel("Discount Percent")
plt.ylabel("Profit")
plt.grid(True)
plt.show()
```

```python
# 20. Customer Rating Analysis
# ============================================================

plt.figure(figsize=(8,5))
plt.hist(df["Customer_Rating"], bins=10)
plt.title("Customer Rating Distribution")
plt.xlabel("Customer Rating")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(8,5))
plt.scatter(df["Customer_Rating"], df["Net_Sales"])
plt.title("Customer Rating vs Net Sales")
plt.xlabel("Customer Rating")
plt.ylabel("Net Sales")
plt.grid(True)
plt.show()
```

```python
# 21. Correlation Analysis
# ============================================================

numeric_df = df.select_dtypes(include=[np.number])
correlation = numeric_df.corr()

print("Correlation Matrix:")
display(correlation)

plt.figure(figsize=(12,8))
plt.imshow(correlation)
plt.colorbar()
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# 22. Save Final Output Files
# ============================================================

df.to_csv("advanced_business_management_analyzed.csv", index=False)
monthly_summary.to_csv("monthly_summary.csv", index=False)
region_summary.to_csv("region_summary.csv", index=False)
product_summary.to_csv("product_summary.csv", index=False)
category_summary.to_csv("category_summary.csv", index=False)
salesperson_summary.to_csv("salesperson_summary.csv", index=False)

print("All analysis completed successfully!")
print("Saved files:")
print("1. advanced_business_management_analyzed.csv")
print("2. monthly_summary.csv")
print("3. region_summary.csv")
print("4. product_summary.csv")
print("5. category_summary.csv")
print("6. salesperson_summary.csv")
```

```python
# 23. Export All Summary Tables to Excel File
# ============================================================

with pd.ExcelWriter("Business_EDA_Report.xlsx") as writer:

    # Original analyzed dataset
    df.to_excel(writer,
                sheet_name="Analyzed_Dataset",
                index=False)

    # Monthly summary
    monthly_summary.to_excel(writer,
                             sheet_name="Monthly_Summary",
                             index=False)

    # Region summary
    region_summary.to_excel(writer,
                            sheet_name="Region_Summary",
                            index=False)

    # Product summary
    product_summary.to_excel(writer,
                             sheet_name="Product_Summary",
                             index=False)

    # Category summary
    category_summary.to_excel(writer,
                              sheet_name="Category_Summary",
                              index=False)

    # Customer type summary
    customer_type_summary.to_excel(writer,
                                   sheet_name="Customer_Type",
                                   index=False)

    # Marketing summary
    marketing_summary.to_excel(writer,
                               sheet_name="Marketing_Channel",
                               index=False)

    # Inventory summary
    inventory_summary.to_excel(writer,
                               sheet_name="Inventory_Status",
                               index=False)

    # Supplier delivery summary
    supplier_delivery.to_excel(writer,
                               sheet_name="Supplier_Delivery",
                               index=False)

    # Salesperson performance
    salesperson_summary.to_excel(writer,
                                 sheet_name="Salesperson_Performance",
                                 index=False)

    # Correlation matrix
    correlation.to_excel(writer,
                         sheet_name="Correlation_Matrix")

print("Excel Report Exported Successfully!")

# ============================================================
# 24. Download Excel File in Google Colab
# ============================================================

from google.colab import files

files.download("Business_EDA_Report.xlsx")

print("Excel file downloaded successfully!")
```
